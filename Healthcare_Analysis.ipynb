# HEALTHCARE ANALYSIS: SLEEP AND LIFESTYLE ANALYSIS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("/content/Sleep_health_and_lifestyle_dataset.csv")
df = data.copy()
df.head()
```

```python
df.isnull().sum()
```

```python
df["Sleep Disorder"] = df["Sleep Disorder"].fillna("Unknown")
df["Sleep Disorder"]
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(df, x = "Heart Rate", y = "Daily Steps")
plt.title("Heart rate based upon daily steps")
plt.ylabel("Daily steps")
plt.xlabel("Heart rate")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(df, x = "Occupation", y = "Stress Level")
plt.title("Stress Level Based upon Occupation")
plt.xlabel("Occupation")
plt.ylabel("Stress Level")
plt.show()
```

```python
df.to_csv("Clean_Healthcare_Data.csv", index=False)
```
